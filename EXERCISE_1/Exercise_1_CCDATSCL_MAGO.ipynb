## Part IV – Copy and View (Lesson 4)

**10. Copy (2 pts)**

Create `original = np.array([1, 2, 3, 4, 5])`. Make a **copy** named `backup`.
Change the 3rd element of `original` to `99`, then print both arrays.
Add a comment explaining why `backup` did or did not change.

````python
original = np.array([1, 2, 3, 4, 5])
backup = original.copy()
original[2] = 99

print("original:", original)
print("backup:", backup)
#  did not change because .copy() creates a new independent array with its own memory data
````

**11. View (2 pts)**

Create `original = np.array([1, 2, 3, 4, 5])`. Make a **view** named `v`.
This time, change the **first element of the view** `v` to `100`, then print both arrays.
Add a comment explaining what happened to `original`.

````python
original = np.array([1, 2, 3, 4, 5])
v = original.view()
v[0] = 100

print("original:", original)
print("view:", v)
# original changed to 100 at index 0 because a view does not own its data; it references the original array's memory.
````

## Part VI – Iterating Arrays (Lesson 6)

**15. Iteration (2 pts)**

- (a) Iterate over `np.array([3, 6, 9])` and print **each element multiplied by 2**.
- (b) Given `grid = np.array([[1, 2, 3], [4, 5, 6]])`, use a **nested for loop** to print every element one by one, then print the **total** of all elements (computed inside the loop).

````python
for x in np.array([3, 6, 9]):
  print(x * 2)
grid = np.array([[1, 2, 3], [4, 5, 6]])
total = 0
for row in grid:
  for val in row:
    print(val)
    total += val
print("Total:", total)
````

## Part XI – Filtering Arrays (Lesson 11)

**24. Filtering (2 pts)**

Given `arr = np.array([12, 45, 67, 23, 89, 50, 34])`:
- (a) Using a **for loop**, build a boolean list `filter_arr` that keeps only values **greater than 40**. Print `filter_arr` and the filtered array.
- (b) Using the **direct (vectorized)** method, return only the **even** elements. Print the boolean mask and the filtered array.

````python
arr = np.array([12, 45, 67, 23, 89, 50, 34])

#a
filter_arr = []
for element in arr:
  if element > 40:
    filter_arr.append(True)
  else:
    filter_arr.append(False)

print(filter_arr)
print(arr[filter_arr])

#b
even_mask = arr % 2 == 0
print(even_mask)
print(arr[even_mask])
````

## Part XII – Random (Lesson 12)

**25. Random Numbers, Distribution and Permutation (2 pts)**

Import `random` from NumPy, then:
- (a) Generate and print a random **integer** from 0 to 49.
- (b) Generate a **2-D array with 2 rows and 4 columns** where each value is `1, 2, or 3` with probabilities `0.2, 0.5, 0.3`. Print it.
- (c) Create `deck = np.array([1, 2, 3, 4, 5])`. Print a random **permutation** of `deck`, then print `deck` to show it was not changed.

*(Outputs will vary every run.)*

````python
from numpy import random

#a
print(random.randint(50))

#b
print(random.choice([1, 2, 3], p=[0.2, 0.5, 0.3], size=(2, 4)))

#c
deck = np.array([1, 2, 3, 4, 5])
print(random.permutation(deck))
print(deck)
````
